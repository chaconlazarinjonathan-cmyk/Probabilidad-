```python
import random
import matplotlib.pyplot as plt
from statistics import mean, median, mode, stdev

# Re-generate fabrica1 to ensure it's populated (from cell ebxwQL2-rAkh)
fabrica1 = []
for i in range(100):
  numero_aleatorio = round ( random.uniform(50, 100), 2)
  fabrica1.append(numero_aleatorio)

# Calculate descriptive statistics for fabrica1 (from cell 3j6D3KCSstzv)
if fabrica1:
    maximo1 = max(fabrica1)
    minimo1 = min(fabrica1)
    rango1 = maximo1 - minimo1
    media1 = mean(fabrica1)
    mediana1 = median(fabrica1)
    try:
        moda1 = mode(fabrica1)
    except Exception:
        moda1 = None # Handle cases where mode might not be found
    desviacion_estandar_fabrica1 = stdev(fabrica1)
else:
    maximo1 = None
    minimo1 = None
    rango1 = None
    media1 = None
    mediana1 = None
    moda1 = None
    desviacion_estandar_fabrica1 = None

plt.figure(figsize = (8,6))
plt.hist(fabrica1, bins=10, alpha=0.2, color='b', edgecolor = "black")

# Only plot axvline if the corresponding statistical value is defined
if media1 is not None:
    plt.axvline(media1, color='b', linestyle='dashed', linewidth=2, label=f'Media={media1:.2f}')
if mediana1 is not None:
    plt.axvline(mediana1, color='r', linestyle='dashed', linewidth=2, label=f'Mediana={mediana1:.2f}')
if moda1 is not None:
    plt.axvline(moda1, color='y', linestyle='dashed', linewidth=2, label=f'Moda={moda1:.2f}')
if media1 is not None and desviacion_estandar_fabrica1 is not None:
    plt.axvline(media1 - desviacion_estandar_fabrica1, color="black", linestyle='dashed', linewidth=2, label=f'Desv. Estandar ={desviacion_estandar_fabrica1:.2f}')
    plt.axvline(media1 + desviacion_estandar_fabrica1, color="black", linestyle='dashed', linewidth=2)

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Produccion diaria de la Fabrica 1")
plt.legend()
plt.show()
```

```python
import random
import matplotlib.pyplot as plt
from statistics import mean, median, mode, stdev

# Re-generate fabrica2 to ensure it's populated (from cell _08MESAusMrV)
fabrica2 = []
for i in range(100):
  numero_aleatorio = round ( random.uniform(70, 100), 2)
  fabrica2.append(numero_aleatorio)

# Calculate descriptive statistics for fabrica2 (from cell eFM1sR9ht6Qk)
# Check if fabrica2 is not empty before calculating statistics
if fabrica2:
    maximo2 = max(fabrica2)
    minimo2 = min(fabrica2)
    rango2 = maximo2 - minimo2
    media2 = mean(fabrica2)
    mediana2 = median(fabrica2)
    try:
        moda2 = mode(fabrica2)
    except Exception: # Handle cases where mode might not be found or data is empty
        moda2 = None
    desviacion_estandar_fabrica2 = stdev(fabrica2)
else:
    # Set variables to None if fabrica2 is empty to prevent errors
    maximo2 = None
    minimo2 = None
    rango2 = None
    media2 = None
    mediana2 = None
    moda2 = None
    desviacion_estandar_fabrica2 = None

# Plotting code
plt.figure(figsize = (8,6))
plt.hist(fabrica2, bins=10, alpha=0.2, color='b', edgecolor = "black")

# Only plot axvline if the corresponding statistical value is defined
if media2 is not None:
    plt.axvline(media2, color='b', linestyle='dashed', linewidth=2, label=f'Media={media2:.2f}')
if mediana2 is not None:
    plt.axvline(mediana2, color='r', linestyle='dashed', linewidth=2, label=f'Mediana={mediana2:.2f}')
if moda2 is not None:
    plt.axvline(moda2, color='y', linestyle='dashed', linewidth=2, label=f'Moda={moda2:.2f}')
if media2 is not None and desviacion_estandar_fabrica2 is not None:
    plt.axvline(media2 - desviacion_estandar_fabrica2, color="black", linestyle='dashed', linewidth=2, label=f'Desv. Estandar ={desviacion_estandar_fabrica2:.2f}')
    plt.axvline(media2 + desviacion_estandar_fabrica2, color="black", linestyle='dashed', linewidth=2)

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Produccion diaria de la Fabrica 2")
plt.legend()
plt.show()
```
